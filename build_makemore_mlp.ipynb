# Build indexes

```python
alphabets = ['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h', 'i', 'j', 'k', 'l', 'm', 'n','o', 'p', 'q', 'r', 's', 't', 'u', 'v', 'w', 'x', 'y','z']
stoi = {s: i+1 for i, s in enumerate(alphabets)}
stoi['.'] = 0
itos = {i: s for s, i in stoi.items()}
```

```python
alphabets = sorted(set(''.join(words)))
stoi = {s: i+1 for i, s in enumerate(alphabets)}
stoi['.'] = 0
itos = {i: s for s, i in stoi.items()}
```

# Building Training Set v01

```python
context = 3
X, Y = [], []
for word in words[:1]:
    word_padded = "..." + word + "."
    for i in range(len(word)+1):
        X.append(word_padded[i:context+i])
        Y.append(word_padded[context+i])
```

```python
for x, y in zip(X,Y):
    print(f"{x} → {y}")
```

# Building Training Set v02

```python
blocksize = 5
X, Y = [], []
for word in words[:1]:
    word_padded = "." * blocksize + word + "."
    for i in range(len(word)+1):
        X.append(word_padded[i:blocksize+i])
        Y.append(word_padded[blocksize+i])
```

```python
for x, y in zip(X, Y):
    print(f"{x} → {y}")
```

# Building Training Set v03

```python
blocksize = 5
X, Y = [], []
for word in words[:5]:
    # print(word)
    context = [0] * blocksize
    for ch in word + '.':
        ix = stoi[ch]
        X.append(context)
        Y.append(stoi[ch])
        context = context[1:] + [ix]

X = torch.tensor(X)
Y = torch.tensor(Y)
```

```python
X.shape, X.dtype, Y.shape, Y.dtype
```

```python
for x, y in zip(X, Y):
    print(f"{x} → {y}")
```
